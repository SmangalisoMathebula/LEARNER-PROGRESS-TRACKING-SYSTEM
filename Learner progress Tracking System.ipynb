```python
from tkinter import Tk, messagebox

root = Tk()
root.withdraw()


class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Learner(Person):
    def __init__(self, learner_id, name, age, course):
        super().__init__(name, age)
        self.learner_id = learner_id
        self.course = course
        self.marks = []
        self.__average_mark = 0

    def set_average(self, average):
        self.__average_mark = average

    def get_average(self):
        return self.__average_mark

    def calculate_average(self):
        try:
            total = recursive_sum(self.marks)
            average = total / len(self.marks)
            self.set_average(average)
            return average
        except ZeroDivisionError:
            return 0

    def get_performance(self):
        average = self.get_average()
        if average < 50:
            return "Fail"
        elif average < 75:
            return "Pass"
        else:
            return "Pass with Distinction"

    def qualifies_for_certificate(self):
        return self.get_average() >= 50


learners = []


def recursive_sum(mark_list):
    if len(mark_list) == 0:
        return 0
    return mark_list[0] + recursive_sum(mark_list[1:])


def learner_exists(learner_id):
    for learner in learners:
        if learner.learner_id == learner_id:
            return True
    return False


def search_learner(learner_id):
    for learner in learners:
        if learner.learner_id == learner_id:
            return learner
    return None


def add_learner():
    try:
        learner_id = input("Enter Learner ID: ")

        if learner_exists(learner_id):
            messagebox.showerror("Error", "Learner ID already exists.")
            return

        name = input("Enter Learner Name: ")
        age = int(input("Enter Learner Age: "))

        if age < 10:
            messagebox.showerror("Invalid Age", "Learner must be at least 10 years old.")
            return

        course = input("Enter Course Name: ")
        learner = Learner(learner_id, name, age, course)
        learners.append(learner)
        messagebox.showinfo("Success", "Learner registered successfully!")

    except ValueError:
        messagebox.showerror("Input Error", "Age must be numeric.")


def view_learners():
    if len(learners) == 0:
        print("\nNo learners found.")
        return

    print("\nLearners")
    for learner in learners:
        print(f"ID: {learner.learner_id} | Name: {learner.name} | Age: {learner.age} | Course: {learner.course}")


def update_learner():
    learner_id = input("Enter Learner ID: ")
    learner = search_learner(learner_id)

    if learner is None:
        messagebox.showerror("Not Found", "Learner not found.")
        return

    learner.name = input("Enter New Name: ")
    learner.course = input("Enter New Course: ")
    messagebox.showinfo("Updated", "Learner updated successfully.")


def delete_learner():
    learner_id = input("Enter Learner ID: ")
    learner = search_learner(learner_id)

    if learner:
        learners.remove(learner)
        messagebox.showinfo("Deleted", "Learner deleted successfully.")
    else:
        messagebox.showerror("Error", "Learner not found.")


def enter_marks():
    learner_id = input("Enter Learner ID: ")
    learner = search_learner(learner_id)

    if learner is None:
        messagebox.showerror("Error", "Learner not found.")
        return

    try:
        num_marks = int(input("How many marks do you want to enter? "))

        for i in range(num_marks):
            try:
                mark = float(input(f"Enter mark {i + 1}: "))

                if mark < 0 or mark > 100:
                    print("Invalid mark. Skipping...")
                    continue

                learner.marks.append(mark)

            except ValueError:
                print("Invalid mark entered.")
                continue

            if len(learner.marks) >= 20:
                print("Maximum marks reached.")
                break

        messagebox.showinfo("Success", "Marks added successfully.")

    except ValueError:
        messagebox.showerror("Input Error", "Enter a valid number.")


def display_marks():
    learner_id = input("Enter Learner ID: ")
    learner = search_learner(learner_id)

    if learner:
        print("\nMarks:", learner.marks)
    else:
        print("Learner not found.")


def show_results(show_message=True):
    learner_id = input("Enter Learner ID: ")
    learner = search_learner(learner_id)

    if learner is None:
        print("Learner not found.")
        return

    average = learner.calculate_average()
    performance = learner.get_performance()
    certificate = "Qualified" if learner.qualifies_for_certificate() else "Not Qualified"

    print("\n========== RESULTS ==========")
    print("Learner ID:", learner.learner_id)
    print("Name:", learner.name)
    print("Course:", learner.course)
    print("Marks:", learner.marks)
    print("Average:", round(average, 2))
    print("Performance:", performance)
    print("Certificate:", certificate)
    print("=============================")

    if show_message:
        messagebox.showinfo(
            "Learner Result",
            f"Average: {average:.2f}\nPerformance: {performance}"
        )


def count_learners(index=0):
    if index >= len(learners):
        return 0
    return 1 + count_learners(index + 1)


def print_menu():
    print("\n========== MENU ==========")
    print("1. Add Learner")
    print("2. View All Learners")
    print("3. Update Learner")
    print("4. Delete Learner")
    print("5. Enter Marks")
    print("6. Display Marks")
    print("7. Search Learner")
    print("8. Show Results")
    print("9. Count Learners")
    print("0. Exit")
    print("==========================")


def main():
    running = True

    while running:
        print_menu()

        try:
            choice = int(input("Enter choice: "))

            if choice == 1:
                add_learner()

            elif choice == 2:
                view_learners()

            elif choice == 3:
                update_learner()

            elif choice == 4:
                delete_learner()

            elif choice == 5:
                enter_marks()

            elif choice == 6:
                display_marks()

            elif choice == 7:
                learner_id = input("Enter Learner ID: ")
                learner = search_learner(learner_id)

                if learner:
                    print("\nLearner Found")
                    print("Name:", learner.name)
                    print("Course:", learner.course)
                else:
                    print("Learner not found.")

            elif choice == 8:
                show_results(show_message=True)

            elif choice == 9:
                print("\nTotal Learners:", count_learners())

            elif choice == 0:
                answer = messagebox.askyesno("Exit", "Are you sure you want to exit?")
                if answer:
                    running = False

            else:
                print("Invalid menu option.")

        except ValueError:
            messagebox.showerror("Menu Error", "Please enter a valid menu number.")

    print("Program closed.")


main()
```
